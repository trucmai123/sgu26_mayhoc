## Bài 4. Slicing và Indexing
Cho biết shape của `a[1]` và `a[1:, 2:4]`. Thử lấy hai phần tử cuối cùng trên chiều thứ ba.

## Bài 5. Reshape và Flatten
Kiểm tra số phần tử trước và sau khi reshape; thử reshape Tensor thành kích thước 2 × 6.

```python
a = torch.randn(3, 4)

print("Original:")
print(a)
print("Number of elements:", a.numel())

print("\nFlatten:")
print(a.ravel())
print("Flatten shape:", a.ravel().shape)

print("\nReshape to 3 x 2 x 2:")
print(a.reshape(3, 2, 2))
print("Reshape shape:", a.reshape(3, 2, 2).shape)

print("\nReshape to 2 x 6:")
print(a.reshape(2, 6))
print("Reshape shape:", a.reshape(2, 6).shape)
```

## Bài 6. Phép toán và nhân ma trận
Phân biệt `a * b` và `a @ b.T`; cho biết shape của kết quả `a @ b.T`.

```python
a = torch.randn(2, 3)
b = torch.randn(2, 3)

print("a:")
print(a)
print("\nb:")
print(b)

print("\na + b:")
print(a + b)

print("\na / b:")
print(a / b)

print("\na^2:")
print(a ** 2)

print("\nElement-wise product a * b:")
print(a * b)

print("\nMatrix multiplication a @ b.T:")
print(a @ b.T)
print("Shape of a @ b.T:", (a @ b.T).shape)
```

## Bài 7. Các hàm thống kê trên Tensor
Thay `dim=0` bằng `dim=1` và nhận xét sự khác nhau.

```python
a = torch.randn(3, 4)

print(a)
print("\nMean (dim=0):")
print(torch.mean(a, dim=0))

print("\nStandard deviation (dim=0):")
print(torch.std(a, dim=0))

print("\nCumulative sum (dim=0):")
print(torch.cumsum(a, dim=0))

print("\nMean (dim=1):")
print(torch.mean(a, dim=1))

print("\nStandard deviation (dim=1):")
print(torch.std(a, dim=1))
```

## Bài 8. Autograd cơ bản
Tính thủ công đạo hàm của `y = x²` tại `x = 3.6` và so sánh với `x.grad`; đổi hàm thành `y = 3*x**2 + 2*x + 1` và kiểm tra gradient.

```python
x = torch.tensor(3.6, requires_grad=True)
y = x * x
y.backward()

print("x =", x)
print("y = x^2 =", y)
print("x.grad =", x.grad)
print("Manual derivative of x^2 at x = 3.6 is:", 2 * 3.6)

x.grad.zero_()
z = 3 * x ** 2 + 2 * x + 1
z.backward()
print("\nGradient of 3*x**2 + 2*x + 1 at x = 3.6:", x.grad.item())
print("Manual derivative: ", 6 * 3.6 + 2)
```

## Bài 9. Gradient Descent tìm hệ số đa thức
So sánh ba hệ số học được với [1, 2, 3]. Giải thích vai trò của `optimizer.zero_grad()`, `mse.backward()` và `optimizer.step()`.

```python
import numpy as np

polynomial = np.poly1d([1, 2, 3])
N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("\nLearned coefficients:")
print(w)
print("Target coefficients:", [1, 2, 3])
```

## Bài 10. Giải hệ phương trình bằng Autograd
Kiểm tra nghiệm thu được với bốn phương trình: `A+B=9`, `C-D=1`, `A+C=8`, `B-D=2` và nhận xét vì sao bài toán có thể có nhiều nghiệm.

```python
import random
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2

    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A.item())
print("B =", B.item())
print("C =", C.item())
print("D =", D.item())
print("\nKiểm tra phương trình:")
print("A + B =", (A + B).item())
print("C - D =", (C - D).item())
print("A + C =", (A + C).item())
print("B - D =", (B - D).item())
```
