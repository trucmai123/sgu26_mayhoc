### Bài tập thực hành
##### Bài 1: Tạo Tensor cơ bản
Tạo một Tensor số nguyên kích thước 2 × 3 và hiển thị kết quả.

Yêu cầu: Cho biết kích thước, số chiều và kiểu dữ liệu của Tensor a.

```python
import torch #thu vien pytorch 

#tao tensor va luu vao bien a
#tensor 2x3
a=torch.tensor(
    [[1, 2, 3],
    [4, 5, 6]],
    dtype=torch.int32 # ep kieu so nguyen 32 bit , dtype thuoc tinh co san
)
print(a)
print(a.shape) #kich thuoc
print(a.ndim) # so chieu
print(a.dtype) #kieu du lieu
```
